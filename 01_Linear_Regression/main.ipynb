# Linear Regression from Scratch

In this notebook we implement Linear Regression using only NumPy.

## Objectives

- Understand the Linear Regression model
- Generate predictions
- Compute Mean Squared Error (MSE)
- Derive the gradients
- Implement Gradient Descent
- Train the model without using scikit-learn

```python
import numpy as np
```

## Dataset

We use a very small dataset to understand the mathematics.

Relationship:

y = 2x + 1

```python
X = np.array([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

Y = np.array([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

print("X Shape :", X.shape)
print("Y Shape :", Y.shape)
```

## Initialize Parameters

Initially the model knows nothing.

So we begin with

- Weight = 0
- Bias = 0

```python
weight = 0.0
bias = 0.0

print(weight)
print(bias)
```

## Prediction

The Linear Regression equation is

Prediction = Weight × X + Bias

```python
predictions = weight * X + bias

print(predictions)
```

## Error

Error = Prediction − Actual

```python
errors = predictions - Y

print(errors)
```

## Mean Squared Error (MSE)

We square every error so that

- Positive and negative errors do not cancel
- Larger errors are penalized more heavily

Loss = Mean(Error²)

```python
loss = np.mean(errors ** 2)

print("Loss =", loss)
```

## Gradient

From the derivation,

Gradient with respect to Weight

dL/dw = (2/N) Σ(Error × X)

Gradient with respect to Bias

dL/db = (2/N) Σ(Error)

```python
gradient_w = (2 / len(X)) * np.sum(errors * X)
gradient_b = (2 / len(X)) * np.sum(errors)

print("Gradient Weight =", gradient_w)
print("Gradient Bias   =", gradient_b)
```

## Gradient Descent

Update Rule

Weight = Weight − Learning Rate × Gradient

Bias = Bias − Learning Rate × Gradient

```python
learning_rate = 0.1

weight -= learning_rate * gradient_w
bias -= learning_rate * gradient_b

print(weight)
print(bias)
```

## Training Loop

Instead of updating the parameters once,

we repeatedly

1. Predict
2. Compute Error
3. Compute Loss
4. Compute Gradients
5. Update Parameters

```python
weight = 0.0
bias = 0.0

learning_rate = 0.1
epochs = 100

for epoch in range(epochs):

    predictions = weight * X + bias

    errors = predictions - Y

    loss = np.mean(errors ** 2)

    gradient_w = (2 / len(X)) * np.sum(errors * X)
    gradient_b = (2 / len(X)) * np.sum(errors)

    weight -= learning_rate * gradient_w
    bias -= learning_rate * gradient_b

    if epoch % 10 == 0:
        print(
            f"Epoch {epoch:3d} | "
            f"Loss={loss:.6f} | "
            f"Weight={weight:.4f} | "
            f"Bias={bias:.4f}"
        )
```

## Final Model

```python
print("Final Weight :", weight)
print("Final Bias   :", bias)
```

```python
predictions = weight * X + bias

print("Predictions")

print(predictions)
```
